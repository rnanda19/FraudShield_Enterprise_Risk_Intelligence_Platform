## Amount & Time distributions (additional EDA, real from your data)

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(df.loc[df["Class"] == 0, "Amount"].clip(upper=500), bins=50, alpha=0.6, label="Legitimate", density=True)
axes[0].hist(df.loc[df["Class"] == 1, "Amount"].clip(upper=500), bins=50, alpha=0.6, label="Fraud", density=True)
axes[0].set_title("Amount distribution (clipped at 500), by class")
axes[0].set_xlabel("Amount")
axes[0].legend()

axes[1].hist(df["Time"] / 3600.0, bins=48, color="steelblue")
axes[1].set_title("Transaction volume by hour of the real time span")
axes[1].set_xlabel("Hours since first transaction")

plt.tight_layout()
plt.savefig("nb01_amount_time_distributions.png", dpi=120)
plt.show()
print("Saved: nb01_amount_time_distributions.png")
```
